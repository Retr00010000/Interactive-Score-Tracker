```python
# Function to append a new integer to the selected list
def List_Add(Int, List):
    List.append(Int)
    print(f"Added {Int}. Current list: {List}")

# Function to remove a specific integer from the selected list safely
def List_Remove(Int, List):
    if len(List) == 0:
        print("The list is empty.")
    else:
        if Int in List:
            List.remove(Int)
            print(f"Removed {Int}. Current list: {List}")
        else:
            print("That score is not in the list.")

# Function to sort the list in descending order (highest to lowest)
def Descending_Sort(List):
    if len(List) > 0:
        List.sort(reverse=True)
        print(f"List sorted in descending order: {List}")
    else:
        print("The list is empty. Nothing to sort.")

# Function to slice and display only the top 3 scores
def Top_Three(List):
    if len(List)>= 3:
        List.sort(reverse=True)
        print(f"Top 3 Scores: {List[:3]}")
    else:
        print("You need at least 3 scores in the list to slice the top 3.")

# Function to find the maximum and minimum values in the list
def Highest_and_Lowest(List):
    if len(List) > 0:
        print(f"Highest Score: {max(List)}")
        print(f"Lowest Score: {min(List)}")
    else:
        print("The list is empty.")

# Function to calculate the mathematical average of the list
def Average(List):
    if len(List) > 0:
        average = sum(List) / len(List)
        print(f"Average Score: {average:.2f}")
    else:
        print("The list is empty.")

# Function to find the very last score (which is the lowest after sorting)
def Last_Score(List):
    if len(List) > 0:
        List.sort(reverse=True)
        print(f"Last Score (Lowest): {List[-1]}")
    else:
        print("The list is empty.")

# Function to combine the list and the tuple, sort them, and print final stats
def All_In_One(List, Tuple):
    if len(List) == 0:
        print("The original list was empty.")

    print(f"Original scores (before tuple): {List}")

    List.extend(Tuple)
    print(f"New scores tuple: {Tuple}\n")

    List.sort(reverse=True)

    print(f"Final scores: {List}")
    print(f"Highest score: {max(List)}")
    print(f"Lowest score: {min(List)}")

    average = sum(List) /len(List)
    print(f"Average score: {average:.2f}")

    print(f"Top 3 scores: {List[:3]}")
    print(f"Last score: {List[-1]}")
    print("\nThank you for using my program.")

# Initialize two empty lists. The second one will become a tuple at the end.
Original_Score=[]
New_Score_Tuple=[]

Name=input("Enter your name Sir/Madam: ")

print(f"\nWelcome {Name}! Both score lists are currently empty. ")

Runner= True

# Main program loop
while Runner:
    Operation_Manager= int(input(f"""
 Which operation would you like to choose for the lists {Name}?
 1. Integer Addition.
 2. Integer Removal.
 3. Descending List Sorting.
 4. List Slicing of the top 3 Scores.
 5. Highest and Lowest Scores.
 6. Average Score.
 7. Last Score.
 8. All In One.

 Enter your choice from (1-8): """))

    print("---------------------------------------------------------------------")

    # Determine which list the user wants to manipulate
    if 1 <= Operation_Manager <= 7:
        target = input("Apply to (1) Original List or (2) New Scores List? : ")
        if target == '2':
            active_list = New_Score_Tuple
        else:
            active_list = Original_Score

    # Match case for user operations
    match Operation_Manager:
        case 1:
            # Try/Except block to catch invalid non-integer inputs
            try:
                score_to_add = int(input("Enter the integer score to add: "))
                List_Add(score_to_add, active_list)
            except ValueError:
                print("Error: Please enter a valid whole number (integer).")

        case 2:
            # Try/Except block to catch invalid non-integer inputs
            try:
                score_to_remove = int(input("Enter the score you want to remove: "))
                List_Remove(score_to_remove, active_list)
            except ValueError:
                print("Error: Please enter a valid whole number (integer).")

        case 3:
            Descending_Sort(active_list)

        case 4:
            Top_Three(active_list)

        case 5:
            Highest_and_Lowest(active_list)

        case 6:
            Average(active_list)

        case 7:
            Last_Score(active_list)

        case 8:
            # Convert the second list to a tuple before final execution
            New_Score_Tuple = tuple(New_Score_Tuple)
            All_In_One(Original_Score, New_Score_Tuple)
            Runner = False

        case _:
            print("Invalid option. Please choose a number from 1 to 8.")

    print("---------------------------------------------------------------------")
```
